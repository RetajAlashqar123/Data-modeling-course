#1. Load the Iris Dataset

Load the Iris dataset, but focus on only the first 100 samples for binary classification as we want to only classify the Setosa and Versicolor classes.

# #2. Split the Dataset into Training and Testing Sets

Note: you need to import train_test_split class.

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris


iris = load_iris()
X, y = iris.data, iris.target



X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

#3. Create and Train the Perceptron Classifier

Configure the perceptron with a maximum of 100 iterations and an initial learning rate of 0.1.

```python
from sklearn.linear_model import Perceptron
clf = Perceptron(max_iter=100,eta0=0.1,random_state=0)
clf.fit(X_train, y_train)
```

**Question #1: Explain what these parameters are and their purposes?**

Let's summarize the parameters for the `Perceptron` classifier:

*   **`max_iter`**: The maximum number of training iterations (epochs). It prevents endless training and sets an upper limit on passes through the data.

*   **`eta0`**: The initial learning rate. This controls the step size taken during weight updates. A smaller `eta0` means slower, potentially more accurate convergence; a larger one means faster, but riskier, convergence.

*   **`random_state`**: An integer seed for the random number generator. This ensures that the initial weights are set consistently, making the training process reproducible.

#4. Use the trained perceptron to make predictions on the test data

```python
y_pred = clf.predict(X_test)
print("Predictions on test data:", y_pred)
```

#5. Generate a Classification Report

```python
from sklearn.metrics import classification_report

# Generate the classification report
report = classification_report(y_test, y_pred)
print("Classification Report:\n", report)
```

**Question #2: Identify the three main steps for implementing a Perceptron in scikit-learn. For each step, specify what you are using (class or method) and briefly explain its purpose.**



The three main steps for implementing a Perceptron in scikit-learn are:

1.  **Initialize (`Perceptron` class)**: Create a `Perceptron` instance, configuring parameters like `max_iter`, `eta0`, and `random_state`.

2.  **Train (`.fit()` method)**: Use the `.fit(X_train, y_train)` method to teach the model the optimal weights and bias from the training data.

3.  **Predict (`.predict()` method)**: Apply the `.predict(X_test)` method to the trained model to get class predictions on new data.
